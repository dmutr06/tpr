### Мета роботи
Навчитися приймати рішення в умовах ризику, коли відомі ймовірності станів природи, за критерієм Байєса–Лапласа.

### Короткі теоретичні відомості

В умовах ризику відомі ймовірності $p_j$ станів природи $O_j$. Для кожної стратегії $x_i$ обчислюють математичне сподівання корисності:

$$E(x_i) = \sum_{j=1}^{n} p_j \, u(x_i, O_j),$$

де $\sum_j p_j = 1$. За критерієм Байєса–Лапласа обирають стратегію з найкращим сподіванням. Тут корисність — це витрати, тому найкраща стратегія та, що дає $\min_i E(x_i)$.

### Постановка задачі

Їдемо до Астани на цілий рік, не знаючи, коли повертаємось. Беремо з собою рівно один з шести комплектів одягу (стратегії Н1–Н6). Перевезення коштує 10 у.о. за кг. Стан природи — місяць повернення, від нього залежить потрібний комплект. Відсутні речі потрібного комплекту докуповуємо на місці за ціною речі + 2 у.о. за кожну.

Вартість $c_{ij}$ = вартість перевезення комплекту $H_i$ + вартість докупівлі в місяці $j$. Корисність — це вартість, тому її мінімізуємо.

### Вихідні дані

```python
import pandas as pd
from IPython.display import display

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
temps = [-17, -17, -10, 3, 13, 18, 20, 18, 12, 3, -7, -14]

print("Середньомісячні температури в Астані, °C")
display(pd.DataFrame([temps], index=["t"], columns=months))

# №: (назва, вага кг, ціна у.о.)
items = {
    1: ("Блайзер", 0.5, 6), 2: ("Зимова куртка", 4, 48), 3: ("Пухові штани", 2, 24),
    4: ("В'єтнамки", 0.5, 6), 5: ("Джинси", 1, 12), 6: ("Кепка", 0.5, 6),
    7: ("Кросівки", 1, 12), 8: ("Куртка", 2, 24), 9: ("Пальто", 3, 36),
    10: ("Рукавички", 0.5, 6), 11: ("Светр", 1, 12), 12: ("Сорочка", 0.5, 6),
    13: ("Футболка", 0.5, 6), 14: ("Черевики", 1.5, 18), 15: ("Чоботи", 2, 24),
    16: ("Шапка", 1, 12), 17: ("Шорти", 0.5, 6),
}
weight = {name: w for name, w, p in items.values()}
price = {name: p for name, w, p in items.values()}

print("Речі")
display(pd.DataFrame(items, index=["Назва", "Вага, кг", "Ціна, у.о."]).T.rename_axis("№"))
```

Комплекти одягу (стратегії):

```python
sets = {
    "Н1": ["Шапка", "Зимова куртка", "Рукавички", "Пухові штани", "Чоботи"],
    "Н2": ["Шапка", "Пальто", "Рукавички", "Джинси", "Чоботи"],
    "Н3": ["Кепка", "Куртка", "Джинси", "Черевики"],
    "Н4": ["Светр", "Джинси", "Кросівки"],
    "Н5": ["Блайзер", "Сорочка", "Джинси", "Кросівки"],
    "Н6": ["Блайзер", "Футболка", "Шорти", "В'єтнамки"],
}
ranges = {"Н1": "< -10", "Н2": "-9..0", "Н3": "1..10", "Н4": "11..20", "Н5": "21..30", "Н6": "30+"}
set_weight = {h: sum(weight[x] for x in names) for h, names in sets.items()}

display(pd.DataFrame({
    "Температура, °C": ranges,
    "Склад комплекту": {h: ", ".join(n) for h, n in sets.items()},
    "Вага, кг": set_weight,
}))
```

Потрібний комплект за температурою: $t \le -10 \to$ Н1, $-9..0 \to$ Н2, $1..10 \to$ Н3, $11..20 \to$ Н4, $21..30 \to$ Н5, $t > 30 \to$ Н6.

**Примітка.** У таблиці значення $-10$ потрапляє в проміжок між «< -10» та «-9..0». Відносимо його до Н1 (березень, $t = -10$).

```python
def needed_set(t):
    if t <= -10: return "Н1"
    if t <= 0: return "Н2"
    if t <= 10: return "Н3"
    if t <= 20: return "Н4"
    if t <= 30: return "Н5"
    return "Н6"

needed = [needed_set(t) for t in temps]
print("Потрібний комплект за місяцями")
display(pd.DataFrame([needed], index=["Комплект"], columns=months))
```

### Побудова матриці вартостей

Функція `purchase_cost` рахує вартість докупівлі, `cost_matrix` будує матрицю 6 комплектів × 12 місяців. Перевіримо її на прикладах з методички: Н1 при потребі Н5 дає $6+6+12+12+2\cdot4 = 44$, Н3 при потребі Н4 дає $12+12+2\cdot2 = 28$.

```python
def purchase_cost(taken, need, prices):
    to_buy = [x for x in sets[need] if x not in sets[taken]]
    return sum(prices[x] + 2 for x in to_buy)

def cost_matrix(prices):
    data = {}
    for h in sets:
        transport = 10 * set_weight[h]
        data[h] = [transport + purchase_cost(h, need, prices) for need in needed]
    return pd.DataFrame(data, index=months).T

assert purchase_cost("Н1", "Н5", price) == 44
assert purchase_cost("Н3", "Н4", price) == 28
print("Перевірка за прикладами з методички пройшла: 44 і 28")
```

### Підзадача 4.2. Рівноймовірні місяці, $p_j = 1/12$

```python
C = cost_matrix(price)

table = C.copy()
table.insert(0, "Перевезення", [10 * set_weight[h] for h in sets])
print("Матриця вартостей, у.о. (аналог табл. 3.5)")
display(table)

print("Що докуповуємо, якщо взяли Н1 (приклад):")
display(pd.DataFrame(
    {m: [", ".join(x for x in sets[n] if x not in sets["Н1"]) or "-"] for m, n in zip(months, needed)},
    index=["Докупити"]))
```

Функція `expected_costs` рахує $E_i = \sum_j p_j c_{ij}$ і показує найкращу стратегію (мінімум, з урахуванням рівних значень).

```python
def expected_costs(C, probs):
    p = pd.Series(probs, index=C.columns)
    assert abs(p.sum() - 1) < 1e-9
    E = (C * p).sum(axis=1)
    best = list(E[abs(E - E.min()) < 1e-9].index)
    return E, best

def show(E, best):
    display(E.round(2).to_frame("E, у.о."))
    print("Найкраща стратегія:", ", ".join(best), "- E =", round(E.min(), 2), "у.о.")

p_uniform = [1 / 12] * 12
E1, best1 = expected_costs(C, p_uniform)
show(E1, best1)
```

### Підзадача 4.3. Імовірності за сезонами

Для кожного сезону ймовірність $1/3$ у трьох місяцях сезону й $0$ в інших.

```python
seasons = {
    "Зима": ["Гру", "Січ", "Лют"],
    "Весна": ["Бер", "Кві", "Тра"],
    "Літо": ["Чер", "Лип", "Сер"],
    "Осінь": ["Вер", "Жов", "Лис"],
}
season_res = {}
for s, ms in seasons.items():
    probs = [1 / 3 if m in ms else 0 for m in months]
    E, best = expected_costs(C, probs)
    season_res[s] = (E, best)
    print(f"{s}: місяці {', '.join(ms)}")
    display(C[ms])
    show(E, best)
```

### Підзадача 4.4. Зима втричі ймовірніша

Вага зимових місяців 3, інших 1, потім нормуємо: сума ваг $3\cdot3 + 9 = 18$.

```python
w = [3 if m in seasons["Зима"] else 1 for m in months]
p_winter = [x / sum(w) for x in w]
print("Ймовірності:", [round(x, 4) for x in p_winter])
E3, best3 = expected_costs(C, p_winter)
show(E3, best3)
```

### Підзадача 4.5. Імовірність пропорційна кількості днів

```python
days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
p_days = [d / 365 for d in days]
E4, best4 = expected_costs(C, p_days)
show(E4, best4)
```

### Підзадача 4.6. Речі №2, 4, 9, 15, 16 дешевші втричі

Ціни цих речей діляться на 3, надбавка +2 у.о. залишається, вага й вартість перевезення не змінюються.

```python
price_cheap = dict(price)
for n in [2, 4, 9, 15, 16]:
    name = items[n][0]
    price_cheap[name] = price[name] / 3
    print(f"{name}: {price[name]} -> {price_cheap[name]:.0f} у.о.")

C_cheap = cost_matrix(price_cheap)
print("Перерахована матриця вартостей, у.о.")
display(C_cheap.round(2))

E5, best5 = expected_costs(C_cheap, p_uniform)
show(E5, best5)
```

### Підсумкова таблиця (аналог табл. 3.7)

```python
rows = {
    "За 12 місяців": (E1, best1),
    "«Зима»": season_res["Зима"],
    "«Весна»": season_res["Весна"],
    "«Літо»": season_res["Літо"],
    "«Осінь»": season_res["Осінь"],
    "«Зима» ×3": (E3, best3),
    "За 12 місяців (за кількістю днів)": (E4, best4),
    "За 12 місяців (зменш. вартість)": (E5, best5),
}
summary = pd.DataFrame(
    {"Найкраща стратегія": [", ".join(b) for E, b in rows.values()],
     "Значення": [round(E.min(), 2) for E, b in rows.values()]},
    index=list(rows))
display(summary)
```

### Висновки

За критерієм Байєса–Лапласа (мінімум сподіваних витрат) в усіх випадках, окрім сезону «Зима», найкращою є стратегія Н4 (светр, джинси, кросівки, 3 кг):

- «Весна»: Н4, 89.33 у.о.; «Літо»: Н4, 30 у.о.; «Осінь»: Н4, 76.67 у.о.;

- ймовірність за кількістю днів: Н4, 87.2 у.о.;

Лише коли повернення гарантовано взимку, найкраще брати Н1 (95 у.о.): його перевезення дорожче, але докуповувати нічого не треба. Н4 підходить для більшості місяців Астани (з травня по вересень потрібен саме він), тому він виграє й коли зима втричі ймовірніша. Подешевшання теплих речей знижує витрати Н4 з 87.5 до 64.83 у.о., бо докупівля взимку стає дешевшою; найкращий вибір при цьому не змінився. Рівних значень, які давали б кілька найкращих стратегій, не було.
